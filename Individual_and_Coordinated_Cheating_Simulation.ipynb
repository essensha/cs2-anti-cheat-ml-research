```python
import pandas as pd

events = pd.DataFrame({
    "player_id": ["A", "A", "A", "B", "B", "B"],
    "reaction_time_ms": [110, 112, 108, 180, 250, 320]
})

features = events.groupby("player_id")["reaction_time_ms"].agg(
    ["mean", "std"]
)

print(features)
```

Creating a Simple Consistency Feature

```python
features["reaction_consistency"] = 1 / (features["std"] + 1)

print(features)
```

```python
 groups = {
    "New": (10, 20000),
    "Experienced": (20, 20000),
    "High-skill": (60, 20000)
}

for group, (fp, legitimate) in groups.items():
    fpr = fp / legitimate
    print(f"{group}: {fpr:.2%}")
```
