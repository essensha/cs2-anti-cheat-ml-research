```python
import pandas as pd

df = pd.read_csv(
    "/srv/Counter Strike cheating.csv",
    skiprows=3
)

print("Dataset shape:", df.shape)
print("\nFirst 10 columns:")
print(df.columns.tolist()[:10])

display(df.head())
```

Explore gameplay event distributions

```python
# Count the frequency of each gameplay event type
event_counts = (
    df["event_type"]
    .value_counts()
    .reset_index()
)

event_counts.columns = ["event_type", "count"]

display(event_counts)

print("Total events:", event_counts["count"].sum())
```

Let's examine whether labeled cheaters actually have fewer footsteps.

```python
footsteps = df[df["event_type"] == "player_footstep"]

print("Available player columns:")
print([c for c in footsteps.columns
       if "steamid" in c or "userid" in c])

display(footsteps.head(3))
```

```python
# Verify the player identifier is populated
print("Missing player IDs:",
      footsteps["user_steamid"].isna().sum())

# Count footsteps per player
footstep_counts = (
    footsteps.groupby("user_steamid")
    .size()
    .reset_index(name="footstep_count")
)

# Add known labels from this match
footstep_counts["cheater"] = (
    footstep_counts["user_steamid"]
    .isin(["Player_3", "Player_4"])
)

display(
    footstep_counts.sort_values(
        "footstep_count", ascending=False
    )
)
```

```python
# Count player spawns
spawns = df[df["event_type"] == "player_spawn"]

spawn_counts = (
    spawns.groupby("user_steamid")
    .size()
    .reset_index(name="spawn_count")
)

# Combine footsteps and spawns
player_features = footstep_counts.merge(
    spawn_counts,
    on="user_steamid",
    how="left"
)

player_features["footsteps_per_spawn"] = (
    player_features["footstep_count"] /
    player_features["spawn_count"]
)

display(
    player_features.sort_values(
        "footsteps_per_spawn",
        ascending=False
    )
)
```

```python
# Identify potential reaction-time and aiming fields
keywords = [
    "time", "tick", "aim", "angle",
    "view", "target", "position",
    "visible", "reaction"
]

candidate_columns = [
    col for col in df.columns
    if any(word in col.lower() for word in keywords)
]

print("Potentially relevant columns:")
print(candidate_columns)
```

```python
shots = df[df["event_type"] == "weapon_fire"].copy()

print("Total weapon-fire events:", len(shots))
print("Missing player IDs:", shots["user_steamid"].isna().sum())

display(
    shots[["user_steamid", "tick", "weapon"]].head(10)
)
```

First few observations include knives and a smoke grenade. We should exclude those before analyzing gunfire timing

Calculate firing intervals

```python
# Exclude knives, grenades, and other non-gun weapons
guns = shots[
    ~shots["weapon"].str.contains(
        "knife|grenade|flashbang|molotov|incgrenade|decoy|c4|taser",
        case=False,
        na=True
    )
].copy()

# Sort shots by player, weapon, and timestamp
guns = guns.sort_values(
    ["user_steamid", "weapon", "tick"]
)

# Calculate time between consecutive shots
guns["shot_interval_ticks"] = (
    guns.groupby(["user_steamid", "weapon"])["tick"]
    .diff()
)

# Summarize intervals by player
interval_summary = (
    guns.groupby("user_steamid")["shot_interval_ticks"]
    .agg(["count", "mean", "std"])
    .reset_index()
)

interval_summary["cheater"] = (
    interval_summary["user_steamid"]
    .isin(["Player_3", "Player_4"])
)

display(interval_summary)
```

```python
# Exclude knives, grenades, and other non-gun weapons
guns = shots[
    ~shots["weapon"].str.contains(
        "knife|grenade|flashbang|molotov|incgrenade|decoy|c4|taser",
        case=False,
        na=True
    )
].copy()

# Sort shots by player, weapon, and timestamp
guns = guns.sort_values(
    ["user_steamid", "weapon", "tick"]
)

# Calculate time between consecutive shots
guns["shot_interval_ticks"] = (
    guns.groupby(["user_steamid", "weapon"])["tick"]
    .diff()
)

# Summarize intervals by player
interval_summary = (
    guns.groupby("user_steamid")["shot_interval_ticks"]
    .agg(["count", "mean", "std"])
    .reset_index()
)

interval_summary["cheater"] = (
    interval_summary["user_steamid"]
    .isin(["Player_3", "Player_4"])
)

display(interval_summary)
```

Both labeled cheaters have relatively high variability. Some legitimate players have much lower variability.

But there's an important methodological problem: This calculation includes long gaps between firefights and rounds. These gaps inflate standard deviations, so we haven't actually measured firing consistency within shooting bursts.

```python
# Explore short intervals between shots
short_intervals = guns[
    guns["shot_interval_ticks"].between(1, 100)
].copy()

summary = (
    short_intervals.groupby("user_steamid")["shot_interval_ticks"]
    .agg(["count", "mean", "std"])
    .reset_index()
)

summary["cheater"] = summary["user_steamid"].isin(
    ["Player_3", "Player_4"]
)

display(summary)
```

```python
weapon_usage = pd.crosstab(
    guns["user_steamid"],
    guns["weapon"]
)

display(weapon_usage)
```

Both players use the M4A1-S, making it a useful starting point for a controlled comparison.

```python
# Restrict analysis to M4A1-S
m4 = guns[
    guns["weapon"] == "weapon_m4a1_silencer"
].copy()

# Calculate intervals between M4A1-S shots
m4 = m4.sort_values(["user_steamid", "tick"])

m4["interval"] = (
    m4.groupby("user_steamid")["tick"].diff()
)

# Focus on short intervals
m4_short = m4[m4["interval"].between(1, 100)]

results = (
    m4_short.groupby("user_steamid")["interval"]
    .agg(["count", "mean", "std"])
    .reset_index()
)

results["cheater"] = results["user_steamid"].isin(
    ["Player_3", "Player_4"]
)

display(results)
```

Inspect round-boundary timestamps

```python
round_events = df[
    df["event_type"].isin([
        "round_prestart",
        "round_freeze_end",
        "round_officially_ended"
    ])
]

display(
    round_events[
        ["event_type", "tick"]
    ].sort_values("tick").head(20)
)
```

Assign rounds to weapon-fire events

```python
import pandas as pd

# Get unique round starting ticks
round_starts = (
    df.loc[
        df["event_type"] == "round_prestart",
        ["tick"]
    ]
    .drop_duplicates()
    .sort_values("tick")
    .reset_index(drop=True)
)

round_starts["round_id"] = range(
    1, len(round_starts) + 1
)

# Assign each gunshot to the most recent round start
guns_by_round = pd.merge_asof(
    guns.sort_values("tick").copy(),
    round_starts,
    on="tick",
    direction="backward"
)

print("Rounds identified:", len(round_starts))
print("Shots without a round:", guns_by_round["round_id"].isna().sum())

display(
    guns_by_round[
        ["user_steamid", "weapon", "tick", "round_id"]
    ].head(10)
)
```

first 10 gunshots have been assigned to Round 1, and their timestamps are in chronological order.

```python
# Select M4A1-S shots
m4_rounds = guns_by_round[
    guns_by_round["weapon"] == "weapon_m4a1_silencer"
].copy()

# Calculate firing intervals within each round
m4_rounds = m4_rounds.sort_values(
    ["user_steamid", "round_id", "tick"]
)

m4_rounds["interval"] = (
    m4_rounds.groupby(
        ["user_steamid", "round_id"]
    )["tick"].diff()
)

# Keep short intervals for exploratory analysis
m4_short = m4_rounds[
    m4_rounds["interval"].between(1, 100)
]

# Summarize by player
results = (
    m4_short.groupby("user_steamid")["interval"]
    .agg(["count", "mean", "std"])
    .reset_index()
)

results["cheater"] = results["user_steamid"].isin(
    ["Player_3", "Player_4"]
)

display(results)
```

Before moving to another feature, let's check whether the original short intervals ever crossed round boundaries

```python
# Original M4A1-S shot sequence
check = guns[
    guns["weapon"] == "weapon_m4a1_silencer"
].copy()

# Assign round IDs
check = pd.merge_asof(
    check.sort_values("tick"),
    round_starts,
    on="tick",
    direction="backward"
)

check = check.sort_values(["user_steamid", "tick"])

# Identify previous shot and previous round
check["previous_tick"] = (
    check.groupby("user_steamid")["tick"].shift()
)

check["previous_round"] = (
    check.groupby("user_steamid")["round_id"].shift()
)

check["interval"] = check["tick"] - check["previous_tick"]

# Count short intervals crossing rounds
cross_round = check[
    check["interval"].between(1, 100)
    & (check["round_id"] != check["previous_round"])
]

print("Short intervals crossing rounds:", len(cross_round))
```

Short intervals crossing rounds: 0 confirms that none of the M4A1-S firing intervals retained by our 100-tick filter crossed a round boundary.

Investigate damage-related features

Our next research question is whether labeled cheaters demonstrate unusual damage patterns relative to legitimate players

```python
# Examine bullet-damage event structure
damage = df[
    df["event_type"] == "bullet_damage"
].copy()

# Identify columns with recorded values
available = damage.columns[
    damage.notna().any()
].tolist()

print("Bullet-damage events:", len(damage))
print("\nAvailable fields:")
print(available)

display(damage[available].head())
```

```python
# Summarize damage-event distances by attacker
distance_summary = (
    damage.groupby("attacker_steamid")["distance"]
    .agg(["count", "mean", "median", "max"])
    .reset_index()
)

distance_summary["cheater"] = (
    distance_summary["attacker_steamid"]
    .isin(["Player_3", "Player_4"])
)

display(
    distance_summary.sort_values(
        "mean",
        ascending=False
    )
)
```

```python
# Count recorded gunfire events per player
shots_per_player = (
    guns.groupby("user_steamid")
    .size()
    .rename("gunfire_events")
)

# Count bullet-damage events per attacker
damage_per_player = (
    damage.groupby("attacker_steamid")
    .size()
    .rename("damage_events")
)

# Combine counts
combat = pd.concat(
    [shots_per_player, damage_per_player],
    axis=1
).fillna(0)

combat["damage_events_per_fire"] = (
    combat["damage_events"] /
    combat["gunfire_events"].replace(0, float("nan"))
)

combat["cheater"] = combat.index.isin(
    ["Player_3", "Player_4"]
)

display(
    combat.sort_values(
        "damage_events_per_fire",
        ascending=False
    )
)
```

Player_3 and Player_4 have relatively high damage-event ratios. But Player_5: a legitimate player with a ratio of 0.444, higher than Player_3.

Test the false-positive tradeoff

```python
# Flag players with high damage-event ratios
combat["flagged"] = combat["damage_events_per_fire"] >= 0.40

# Compare flags against known labels
print(pd.crosstab(
    combat["cheater"],
    combat["flagged"],
    rownames=["Actual cheater"],
    colnames=["Flagged suspicious"]
))
```

###Build our player-level feature dataset


Before training a model, we need one row per player, with all three features and the known cheating labels

```python
# Start with our combat feature table
features = combat[
    ["damage_events_per_fire", "cheater"]
].copy()

# Add footsteps per spawn
footstep_feature = (
    player_features
    .set_index("user_steamid")["footsteps_per_spawn"]
)

features = features.join(footstep_feature)

# Add M4A1-S firing interval variability
firing_feature = (
    results
    .set_index("user_steamid")["std"]
    .rename("m4_firing_std")
)

features = features.join(firing_feature)

print("Dataset shape:", features.shape)
display(features)
print("\nMissing values:")
print(features.isna().sum())
```

 Create a model-ready baseline

```python
# Select complete features for the initial baseline
baseline_features = features[
    ["damage_events_per_fire", "footsteps_per_spawn", "cheater"]
].copy()

# Separate predictors from labels
X = baseline_features.drop(columns="cheater")
y = baseline_features["cheater"].astype(int)

print("Feature matrix:", X.shape)
print("Labels:", y.value_counts().to_dict())

print("\nMissing values:")
print(X.isna().sum())

display(X)
```

```python
import pandas as pd

labels = pd.read_csv(
    "/srv/cs2_match_0_cheater_labels.csv"
)

print("Shape:", labels.shape)
print("Columns:", labels.columns.tolist())

display(labels)
```

```python
# Prepare the source labels
verified_labels = labels.set_index("steamid")[
    "is_labeled_cheater"
]

# Merge source labels with engineered features
model_df = X.join(verified_labels)

print("Dataset shape:", model_df.shape)
print("Missing labels:", model_df["is_labeled_cheater"].isna().sum())

display(model_df)
```
